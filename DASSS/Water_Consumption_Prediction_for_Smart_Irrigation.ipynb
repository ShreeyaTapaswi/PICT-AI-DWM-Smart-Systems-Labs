### Missing Values


#### Identifying Null values

```python
# count per column
df.isnull().sum()
```

```python
# Percentage per column
missing_pct = df.isnull().sum()/len(df)*100
missing_pct
```

#### Visualization of Missing Values

```python
# Plot the missing data - Bar Graph
plt.figure(figsize=(8, 4))
missing_pct.plot(kind="bar", color="skyblue")
plt.ylabel("Percentage Missing (%)")
plt.title("Missing Values Per Column")
plt.show()
```

#### Handle Null values

```python
# Fill Numerical columns with mean
for col in numerical_columns:
    mean_value = df[col].mean()
    df[col]=df[col].fillna(mean_value)
```

```python
# Fill Categorical columns with mode
for col in categorical_columns:
    mode_value = df[col].mode()[0]
    df[col]=df[col].fillna(mode_value)
```

```python
# Drop the columns with too many missing values (>70%)
for col in df.columns:
    if missing_pct[col] > 50:
        print(f"\nDropping column '{col}' (> 50% missing)")
        df = df.drop(columns=[col])
        if col in numerical_columns:
            numerical_columns.remove(col)
        if col in categorical_columns:
            categorical_columns.remove(col)
```

### Outliers

#### Identify Outliers

```python
print("Column wise outliers count: ")
for col in numerical_columns:
    q1 = df[col].quantile(0.25)
    q3 = df[col].quantile(0.75)
    iqr = q3-q1

    lower_bound = q1 - 1.5*iqr
    upper_bound = q3 + 1.5*iqr

    outliers = df[(df[col]<lower_bound)|(df[col]>upper_bound)]
    outliers_count = outliers.shape[0]
    print(f"{col}: {outliers_count} outliers")
```

#### Visualize Outliers uding Box Plot

```python
# Plot a boxplot for each numerical column
for col in numerical_columns:
    plt.figure(figsize=(6, 2))
    sns.boxplot(x=df[col])
    plt.title(f"Outliers in {col}")
    plt.show()
```

```python
# Pairplot can crash if there are raw missing values; drops them just for the plot
sns.pairplot(df[numerical_columns].dropna())
plt.suptitle("Pairplot of Numerical Columns", y=1.02)
plt.show()
```

```python
# Plot distributions for all numerical columns
for col in numerical_columns:
    plt.figure(figsize=(6, 3))
    sns.histplot(df[col].dropna(), kde=True, color="purple")
    plt.title(f"Distribution of {col}")
    plt.show()
```

```python
plt.figure(figsize=(6, 4))
# Calculates correlation scores between -1 and 1
sns.heatmap(df[numerical_columns].corr(), annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Correlation Matrix Heatmap")
plt.show()
```

```python
# Create a pie chart for each categorical column
for col in categorical_columns:
    # Get the value counts of each unique category
    category_counts = df[col].value_counts()

    plt.figure(figsize=(5, 5))
    plt.pie(
        category_counts,
        labels=category_counts.index,
        autopct="%1.1f%%",
        startangle=140,
        colors=sns.color_palette("Set2"),
    )
    plt.title(f"Category Breakdown: {col}")
    plt.show()
```
