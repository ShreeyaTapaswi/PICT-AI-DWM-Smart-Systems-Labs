### **Step 4: Handle Missing Values**

Fill missing numerical values using mean.

Fill missing categorical values using mode.

Drop column having too many missing values.

```python
df['age'].fillna(df['age'].mean(), inplace=True)
```

```python
df['embarked'].fillna(df['embarked'].mode()[0], inplace=True)
```

```python
df['embark_town'].fillna(df['embark_town'].mode()[0], inplace=True)
```

```python
df.drop(columns=['deck'], inplace=True)
```

```python
df.isnull().sum()
```

Explanation
Mean is used for numerical columns.
Mode is used for categorical columns.
Columns with excessive missing values are removed.
Deck columns had 688 null values out of 819 total.

### **Step 5: Convert Data Types**

```python
#Convert float to int.
df['age'] = df['age'].astype(int)

# Convert categorical integer column to category
df['pclass'] = df['pclass'].astype('category')

# Convert object columns to category
df['sex'] = df['sex'].astype('category')
df['embarked'] = df['embarked'].astype('category')
df['class'] = df['class'].astype('category')
df['who'] = df['who'].astype('category')
df['alive'] = df['alive'].astype('category')

# Boolean to integer
df['adult_male'] = df['adult_male'].astype(int)
```

Data types were examined using df.dtypes. Variables already stored in appropriate formats were left unchanged. Categorical variables (object) were converted to the category data type to improve memory efficiency and prepare them for encoding. Boolean variables were converted to integers where required for machine learning. Integer variables representing counts or class labels were retained as integers because they were already in the correct format.

```python
df.dtypes
```

Explanation

Sometimes variables are stored in incorrect formats.

Using astype() converts them into appropriate data types.

### **Step 6: Convert Categorical Variables into Numerical Variables**

Using Label Encoding

```python
from sklearn.preprocessing import LabelEncoder

# Ensure clean, correct values before encoding
df['embarked'] = df['embarked'].astype(str)

le = LabelEncoder()

df['sex'] = le.fit_transform(df['sex'])
df['embarked'] = le.fit_transform(df['embarked'])
df['class'] = le.fit_transform(df['class'])
df['who'] = le.fit_transform(df['who'])
df['alive'] = le.fit_transform(df['alive'])
```

Machine learning algorithms cannot process text.

Label Encoding converts text into numbers.

### **Step 7: Data Formatting**

```python
#Check duplicate records.
print(df.duplicated().sum())
```

```python
#Remove duplicates.
df = df.drop_duplicates()
```

Explanation

### **Step 8: Data Normalization**

```python
#Normalize numerical columns.
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()

df[['age', 'fare']] = scaler.fit_transform(df[['age', 'fare']])
```

Explanation

Normalization scales values between 0 and 1.

Formula

(x − min) / (max − min)
